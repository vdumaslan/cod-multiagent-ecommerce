```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from pathlib import Path
```

```python
# Set random seed for reproducibility
np.random.seed(42)

print("Data Split Preparation Script")
print("Train/Validation/Test Splits for Multi-Agent System")
```

# For Product Intelligence Agent

```python
# Load cleaned product metadata
META_PATH = Path("../data/processed/amazon_product_meta/amazon_product_meta_cleaned.parquet")
df_meta = pd.read_parquet(META_PATH)

# Display the first few rows of the dataframe
print(df_meta.head())
```

```python
# Split Product Intelligence Data (80/10/10 Stratified)
train_val_df, test_df = train_test_split(
    df_meta,
    test_size=0.1,
    stratify=df_meta['category'],
    random_state=42
)

train_df, val_df = train_test_split(
    train_val_df,
    test_size=0.1111,  # 0.1111 x 0.9 ≈ 0.1 of the total data
    stratify=train_val_df['category'],
    random_state=42
)

# Print the sizes of each split
print(f"Training Set Size: {len(train_df)}")
print(f"Validation Set Size: {len(val_df)}")
print(f"Test Set Size: {len(test_df)}")
```

```python
# Verify stratification
print("Category Distribution in Training Set:")
print(train_df['category'].value_counts(normalize=True))
print("Category Distribution in Validation Set:")
print(val_df['category'].value_counts(normalize=True))
print("Category Distribution in Test Set:")
print(test_df['category'].value_counts(normalize=True))
```

```python
# Samples from Product Intelligence Agent

print("----- Product Intelligence TRAIN sample -----")
print(train_df.sample(5))
print(f"\nShape: {train_df.shape}\n")
print("----- Product Intelligence VAL sample -----")
print(val_df.sample(5))
print(f"\nShape: {val_df.shape}\n")
print("----- Product Intelligence TEST sample -----")
print(test_df.sample(5))
print(f"\nShape: {test_df.shape}\n")
```

```python
# Save the splits to CSV files
OUTPUT_DIR = Path("../data/processed/data_splits")
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

train_df.to_parquet(OUTPUT_DIR / "product_intelligence_train.parquet", index=False)
val_df.to_parquet(OUTPUT_DIR / "product_intelligence_val.parquet", index=False)
test_df.to_parquet(OUTPUT_DIR / "product_intelligence_test.parquet", index=False)

print("Data splits saved successfully.")
```
