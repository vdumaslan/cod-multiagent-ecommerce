```python
from prefect import task, flow

@task
def ingest_raw():
    print("Ingestion step completed.")

@task
def clean_transform():
    print("Cleaning and transformation completed.")

@task
def compute_embeddings():
    print("Embedding computation completed.")

@task
def update_faiss():
    print("FAISS index update completed.")

@task
def run_sentiment():
    print("Sentiment agent step completed.")

@task
def run_agents():
    print("Ranking, pricing, and orchestrator agents run successfully.")
```

```python
@flow(name="ecommerce_pipeline_flow")
def ecommerce_pipeline_flow():
    ingest_raw()
    clean_transform()
    compute_embeddings()
    update_faiss()
    run_sentiment()
    run_agents()
    print("Pipeline completed successfully.")
```

```python
ecommerce_pipeline_flow()
```
